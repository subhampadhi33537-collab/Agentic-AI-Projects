```python
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain.agents import create_agent
from langchain_community.tools import DuckDuckGoSearchRun


from langgraph.graph import StateGraph,START,END

from dotenv import load_dotenv
from typing import TypedDict
```

```python
load_dotenv()
```

```python
model=ChatGroq(model="openai/gpt-oss-120b")

prompt=PromptTemplate(
        template="""You are a helpful AI research agent.
Your job is to answer the user's question accurately and concisely.
Tools:
- DuckDuckGo Search: Use this tool when the question requires current, factual, or web-based information.
Rules:
1. Understand the user's query first.
2. Use DuckDuckGo Search when up-to-date information is needed.
3. Use the retrieved information to formulate the answer.
4. Do not make up facts or sources.
5. If reliable information cannot be found, clearly say so.
6. Keep answers short, clear, and useful.
User Query:\n {user_input}""",
input_variables=["user_input"]
)


search_tool=DuckDuckGoSearchRun()
agent=create_agent(
    model=model,
    tools=[search_tool]
)
```

```python
class ChatState(TypedDict):
    user_message:str
    ai_message:str
```

```python
def llm_response(state: ChatState):
    query = state["user_message"]
    result = agent.invoke({
        "messages": [
            {
                "role": "user",
                "content": query
            }
        ]
    })
    return {"ai_message": result["messages"][-1].content}
```

```python
graph=StateGraph(ChatState)

graph.add_node("model",llm_response)

graph.add_edge(START,"model")
graph.add_edge("model",END)

chatbot=graph.compile()
```

```python
chatbot
```

```python
while True:
    user_input=input()
    if user_input  in ["stop","EXIT","STOP","exit"]:
        break
    result=chatbot.invoke({"user_message":user_input})
    print(result["ai_message"])
```
